# Friedman test
H0: all models perform the same. Unit of analysis is the dataset: scores are averaged over model seeds first.

```python
from pathlib import Path

import pandas as pd
from scipy.stats import friedmanchisquare

metric = "balanced_accuracy"  # "balanced_accuracy", "f1", "roc_auc"
deep_forests = ["bare", "ifc", "projection", "combined"]

df = pd.read_csv(Path("../results") / "all_tabular_results.csv")
```

```python
def friedman(df, params, models):
    table = df[df["params"] == params].pivot_table(index="dataset", columns="model", values=metric)[models]
    stat, p = friedmanchisquare(*[table[m] for m in models])
    ranks = table.rank(axis=1, ascending=False).mean().sort_values()
    print(f"{params}, {len(models)} models, {len(table)} datasets: chi2 = {stat:.3f}, p = {p:.4f}")
    print("mean rank (1 = best):")
    print(ranks.round(2).to_string(), "\n")
```

```python
friedman(df, "default", deep_forests)
friedman(df, "tuned", deep_forests)
```
